```python
from enb import Classifier, create_dataset_from_json

# dataset_path = "data/fake_newsgroup.json"
dataset_path = "data/20_newsgroups.json"
train, test = create_dataset_from_json(dataset_path)

c = Classifier(set(t["label"] for t in train))
for t in train:
    c.train(t["label"], t["text"])
```

```python
stats = c.explain(test[0]["text"])
stats.print_to_output()
```
